```python
import ast

def convert(text):
    l = []
    for i in ast.literal_eval(text):
        l.append(i['name'])
    return l           
```

```python
merged_data['genres'] = merged_data['genres'].apply(convert)
```

```python
import ast

def convert_cast(text):
    l = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 4:
            l.append(i['name'])
        counter += 1
    return l    
```

```python
merged_data['cast'] = merged_data['cast'].apply(convert_cast)
```

```python
import ast

def fectch_director(text):
    l = []
    counter = 0
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l    
```

```python
merged_data['crew'] = merged_data['crew'].apply(fectch_director)
```

```python
merged_data['overview'] = merged_data['overview'].apply(lambda x:x.split())
```

```python
def remove_space(word):
    l = []
    for i in word:
        l.append(i.replace(" ", ""))
    return l
```

```python
merged_data['cast'] = merged_data['cast'].apply(remove_space)
```

```python
merged_data.head(5)
```

```python
merged_data['crew'] = merged_data['crew'].apply(remove_space)
```

```python
merged_data['tags'] = merged_data['genres'] + merged_data['keywords'] + merged_data['overview'] + merged_data['cast'] + merged_data ['crew']
```

```python
new_df = merged_data[['movie_id', 'title', 'tags']]
```

```python
new_df.head(5)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_df.head(5)
```

```python
new_df.iloc[0]['tags']
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head(5)
```

```python
new_df.iloc[0]['tags']
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stems(text):
    l = []
    for i in text.split():
        l.append(ps.stem(i))
    return " " .join(l)
```

```python
new_df['tags'] = new_df['tags'].apply(stems)
```

```python
new_df.iloc[0]['tags']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000, stop_words = 'english')
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
similarity.shape
```

```python

from difflib import get_close_matches

def recommend(movie):
    titles = new_df['title'].tolist()

    matches = get_close_matches(movie, titles, n=1, cutoff=0.6)
    if not matches:
        print(f"No close match found for '{movie}'")
        return

    matched_title = matches[0]
    if matched_title != movie:
        print(f"Showing results for: '{matched_title}'\n")

    index = new_df.index.get_loc(new_df[new_df['title'] == matched_title].index[0])
    distance = sorted(enumerate(similarity[index]), reverse=True, key=lambda x: x[1])
    for i in distance[1:6]:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Avengers')
```
